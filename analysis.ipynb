```python
# 📦 1. Import Library & Load Data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Global_Cybersecurity_Threats_2015-2024.csv")
```

```python
# 📊 2. Year-over-Year Threat Trend
plt.figure(figsize=(12,6))
sns.countplot(x="Year", hue="Year", data=df, palette="Blues_d")
plt.title("Total Number of Cybersecurity Incidents Per Year")
plt.xlabel("Year")
plt.ylabel("Number of Incidents")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

📊 1. Year-over-Year Threat Trend
Caption: Grafik ini menunjukkan jumlah total insiden keamanan siber global dari tahun 2015 hingga 2024. Terlihat tren peningkatan signifikan dalam jumlah serangan selama dekade terakhir, menyoroti urgensi peningkatan kesiapsiagaan keamanan digital.



🌍 2. Top Affected Countries
Caption: Negara-negara dengan jumlah insiden keamanan siber terbanyak ditampilkan dalam grafik ini. Negara-negara dengan infrastruktur digital besar dan ekonomi kuat cenderung menjadi target utama serangan.

```python
# 🔍 3. Attack Type Frequency
plt.figure(figsize=(12,6))
sns.countplot(
    y="Attack Type", 
    data=df, 
    order=df["Attack Type"].value_counts().index, 
    hue="Attack Type", 
    palette="mako", 
    legend=False
)
plt.title("Most Common Types of Cyber Attacks")
plt.xlabel("Number of Incidents")
plt.ylabel("Attack Type")
plt.tight_layout()
plt.show()
```

🔍 3. Attack Type Frequency
Caption: Grafik ini menggambarkan distribusi frekuensi dari berbagai jenis serangan siber. Jenis serangan seperti phishing dan ransomware merupakan ancaman dominan selama periode 2015–2024.

```python
# 💸 4. Financial Impact by Attack Type
impact_by_attack = df.groupby("Attack Type")["Financial Loss (in Million $)"].mean().sort_values(ascending=False)
plt.figure(figsize=(12,6))
impact_by_attack.plot(kind='bar', color='crimson')
plt.title("Average Financial Loss by Attack Type")
plt.ylabel("Financial Loss (Million $)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

💸 4. Financial Impact by Attack Type
Caption: Rata-rata kerugian finansial akibat tiap jenis serangan siber. Beberapa jenis serangan tertentu, meskipun tidak sering terjadi, menyebabkan dampak ekonomi yang jauh lebih besar.

```python
# 🏭 5. Most Targeted Industries
plt.figure(figsize=(10,6))
sns.countplot(
    y="Target Industry", 
    data=df, 
    order=df["Target Industry"].value_counts().index, 
    hue="Target Industry", 
    palette="coolwarm", 
    legend=False
)
plt.title("Most Targeted Industries")
plt.xlabel("Number of Incidents")
plt.ylabel("Industry")
plt.tight_layout()
plt.show()
```

🏭 5. Most Targeted Industries
Caption: Industri yang paling sering menjadi sasaran serangan siber ditampilkan di sini. Sektor seperti IT, layanan kesehatan, dan keuangan merupakan target utama karena data sensitif dan nilai ekonominya.

```python
# 🕵️ 6. Attack Source Patterns
attack_sources = df["Attack Source"].value_counts()
plt.figure(figsize=(8,5))
attack_sources.plot(kind='pie', autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title("Distribution of Cyber Attack Sources")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

🕵️ 6. Attack Source Patterns
Caption: Distribusi sumber serangan siber. Kelompok peretas (hacker group) dan aktor negara (nation-state) merupakan sumber dominan, menandakan campuran ancaman kriminal dan geopolitik.

```python
# ⏱️ 7. Resolution Time by Year
plt.figure(figsize=(12,6))
sns.boxplot(
    x="Year", 
    y="Incident Resolution Time (in Hours)", 
    data=df, 
    hue="Year", 
    palette="Set2", 
    legend=False
)
plt.title("Distribution of Incident Resolution Time per Year")
plt.ylabel("Resolution Time (Hours)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

⏱️ 7. Resolution Time by Year
Caption: Distribusi waktu penyelesaian insiden dari tahun ke tahun. Menunjukkan seberapa cepat organisasi dapat merespons dan memulihkan sistem setelah insiden terjadi.

```python
# 👥 8. User Impact by Attack Type
impact_users = df.groupby("Attack Type")["Number of Affected Users"].mean().sort_values(ascending=False)
plt.figure(figsize=(12,6))
impact_users.plot(kind='bar', color='orange')
plt.title("Average Number of Users Affected by Attack Type")
plt.ylabel("Number of Affected Users")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

👥 8. User Impact by Attack Type
Caption: Rata-rata jumlah pengguna yang terdampak untuk tiap jenis serangan. Menunjukkan potensi skala kerusakan pada pengguna akhir berdasarkan taktik yang digunakan oleh penyerang.

```python
# Rata-rata kerugian finansial per jenis pertahanan
avg_loss = df.groupby("Defense Mechanism Used")["Financial Loss (in Million $)"].mean().sort_values()

# Rata-rata waktu penyelesaian insiden per jenis pertahanan
avg_resolution_time = df.groupby("Defense Mechanism Used")["Incident Resolution Time (in Hours)"].mean().sort_values()

# Plot Financial Loss
plt.figure(figsize=(14,5))
plt.subplot(1, 2, 1)
avg_loss.plot(kind='barh', color='teal')
plt.title("Avg Financial Loss by Defense Mechanism")
plt.xlabel("Financial Loss (Million $)")
plt.ylabel("Defense Mechanism")

# Plot Resolution Time
plt.subplot(1, 2, 2)
avg_resolution_time.plot(kind='barh', color='coral')
plt.title("Avg Resolution Time by Defense Mechanism")
plt.xlabel("Resolution Time (Hours)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```


Caption:Dua grafik ini membandingkan efektivitas berbagai mekanisme pertahanan siber berdasarkan rata-rata kerugian finansial dan waktu penyelesaian insiden. Mekanisme yang lebih efektif ditandai dengan kerugian yang lebih rendah dan waktu resolusi yang lebih cepat.
